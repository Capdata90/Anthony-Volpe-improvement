```python
import pandas as pd
import numpy as np

# Usa el df que ya tienes cargado
x_bins = np.linspace(-1.5, 1.5, 20)
z_bins = np.linspace(1.0, 4.5, 20)

df["xbin"] = pd.cut(df["plate_x"], bins=x_bins, labels=False)
df["zbin"] = pd.cut(df["plate_z"], bins=z_bins, labels=False)

grid = df.groupby(["xbin","zbin"]).agg(
    swings=("is_swing","sum"),
    whiffs=("is_whiff","sum")
).reset_index()
grid["whiff_pct"] = (grid["whiffs"] / grid["swings"].clip(lower=1) * 100).round(1)

print(grid[grid["swings"] > 0].head(20))
print(f"\nTotal celdas con datos: {len(grid[grid['swings'] > 0])}")
```

```python
import pandas as pd
import numpy as np
from pybaseball import statcast_batter

# Cargar datos
df = statcast_batter('2025-03-20', '2025-04-21', player_id=683011)

# Crear columnas
df["is_whiff"] = df["description"].isin(["swinging_strike", "swinging_strike_blocked"]).astype(int)
df["is_swing"] = df["description"].isin([
    "swinging_strike", "swinging_strike_blocked",
    "foul", "foul_tip", "hit_into_play"
]).astype(int)

# Verificar heatmap
x_bins = np.linspace(-1.5, 1.5, 20)
z_bins = np.linspace(1.0, 4.5, 20)

df = df.dropna(subset=["plate_x", "plate_z"])
df["xbin"] = pd.cut(df["plate_x"], bins=x_bins, labels=False)
df["zbin"] = pd.cut(df["plate_z"], bins=z_bins, labels=False)

grid = df.groupby(["xbin","zbin"]).agg(
    swings=("is_swing","sum"),
    whiffs=("is_whiff","sum")
).reset_index()
grid["whiff_pct"] = (grid["whiffs"] / grid["swings"].clip(lower=1) * 100).round(1)

print(f"Celdas con datos: {len(grid[grid['swings'] > 0])}")
print(grid[grid["swings"] > 0].head(10))
```
